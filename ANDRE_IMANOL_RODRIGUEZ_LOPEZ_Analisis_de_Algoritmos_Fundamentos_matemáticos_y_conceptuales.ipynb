- *¿Qué establece el Principio de Invarianza respecto al tiempo de ejecución de dos implementaciones distintas para un mismo algoritmo?*  
**RESPUESTA:**  
El **Principio de Invarianza** es un fundamento matemático en el análisis de algoritmos que establece que **el tiempo de ejecución de dos implementaciones distintas para un mismo algoritmo no difiere más que en una constante multiplicativa**.

Formalmente, si un algoritmo tiene dos implementaciones (\\(I_1\\) e \\(I_2\\)) cuyos tiempos de ejecución son \\(T_1(n)\\) y \\(T_2(n)\\) para una entrada de tamaño \\(n\\), existen constantes positivas \\(c\\) y \\(n_0\\) tales que para cualquier entrada suficientemente grande (\\(n \ge n_0\\)) se cumple que:

\\[T_1(n) \le c \cdot T_2(n)\\]

Esta constante multiplicativa se conoce en la literatura como la **"constante oculta"**. La consecuencia directa de este principio es que **la eficiencia de un algoritmo depende de su propia naturaleza y diseño**, y no de factores externos o coyunturales de la implementación. Factores como la potencia física de la computadora, el compilador seleccionado, el lenguaje de programación elegido o la destreza del programador solo afectan a esa constante multiplicativa \\(c\\), pero no alteran el orden de complejidad asintótica del algoritmo.

#### 1. Funciones de piso (*floor*) y techo (*ceiling*) en computación
Las funciones de **piso** (\\(\lfloor x \rfloor\\)) y **techo** (\\(\lceil x \rceil\\)) se utilizan en computación para transformar números reales o de punto flotante en valores enteros bajo reglas específicas:
*   **Definición:** El piso de un número real \\(x\\) es el entero más grande que es menor o igual que \\(x\\). El techo de \\(x\\) es el entero más pequeño que es mayor o igual que \\(x\\). Ambas funciones son monótonas crecientes.
*   **Mínimo de bits de codificación:** Se utiliza la función techo para calcular el **número mínimo de bits necesarios** para codificar una colección de \\(n\\) objetos únicos, mediante la fórmula \\(\lceil \log_2 n \rceil\\). Por ejemplo, para almacenar 1000 códigos se requieren al menos \\(\lceil \log_2 1000 \rceil = 10\\) bits.
*   **División de problemas y recursión:** En el análisis de algoritmos que dividen un problema en subproblemas (como en la técnica de *Divide y Vencerás* para búsquedas recursivas o mergesort), el tamaño de las divisiones no siempre da un resultado entero; por ello, se emplean pisos y techos para mantener las dimensiones de los subproblemas dentro del dominio de los enteros (por ejemplo, \\(\lfloor n/2 \rfloor\\) o \\(\lceil n/2 \rceil\\)).
*   **Aritmética modular:** La función piso es fundamental para definir formalmente el residuo o módulo (\\(a \bmod n\\)) de una división entera, expresado matemáticamente bajo la identidad: \\(a \bmod n = a - n \lfloor a/n \rfloor\\).

#### 2. Por qué la exponencial crece más rápido que cualquier polinomio
Según el texto, para cualquier constante real \\(b\\) y cualquier base constante \\(a > 1\\), se cumple el siguiente límite matemático fundamental cuando el tamaño de la entrada \\(n\\) tiende al infinito:

\\[\lim_{n \to \infty} \frac{n^b}{a^n} = 0\\]

A partir de este límite, se concluye formalmente en el análisis asintótico que el polinomio es de un orden estrictamente menor que la exponencial:

\\[n^b = o(a^n)\\]

Esto demuestra de manera matemática que **cualquier función exponencial positiva crece estrictamente más rápido que cualquier función polinomial**.

En la práctica computacional, esta diferencia es de suma importancia: mientras que los algoritmos de complejidad polinomial (\\(n^b\\)) se consideran **tratables y eficientes** en la práctica para grandes volúmenes de datos, los algoritmos de complejidad exponencial (\\(a^n\\)) sufren de una **explosión combinatoria** que los vuelve **intratables e imprácticos**, requiriendo siglos de cómputo para resolver problemas con tamaños de entrada medianos.

#### 1. Las funciones de Techo y Piso
En nuestra vida diaria usamos números con decimales (como 2.5), pero en las computadoras muchas cosas físicas o lógicas **tienen que ser números enteros completos**. No puedes tener "medio bit" de memoria ni dar "tres vueltas y media" a un ciclo de forma física. Para resolver esto, las computadoras usan estas dos funciones de redondeo:

En computación, esto se usa para saber exactamente cuántos recursos mínimos enteros (como bits o bloques de memoria) necesitas para guardar una cantidad de datos.
*   **La función Piso (redondea siempre para abajo):** Imagina que tienes \$11 de propina para repartir en partes iguales entre tus 5 amigos sin tener que romper los billetes. Al dividir \\(11 / 5\\) te da \\(\$2.20\\) por persona. Lo justo es darle **\$2** a cada uno y dejar el resto. El piso simplemente "borra" los decimales e ignora lo que sobra. En computación se usa para divisiones exactas y operaciones de residuo.

#### 2. La carrera entre el Polinomio y la Exponencial
Para entender por qué una exponencial le gana siempre a un polinomio, imagina que dos personas comienzan a ahorrar dinero desde el Día 1 de la siguiente manera:

*   **El Polinomio (por ejemplo, ahorrar el día al cuadrado: \\(n^2\\) dólares):** Tu ahorro es el número del día multiplicado por sí mismo.
    *   El día 2 ahorras: \\(2 \times 2 = \$4\\).
    *   El día 10 ahorras: \\(10 \times 10 = \$100\\).
    *   Crece a un ritmo constante y potente.
*   **La Exponencial (por ejemplo, duplicar tu dinero: \\(2^n\\) dólares):** Tu ahorro se duplica con respecto al día anterior.
    *   El día 2 ahorras: \$4.
    *   El día 10 ahorras: **\$1,024**.
    *   El día 30 ahorras: **¡Más de mil millones de dólares!**

Al inicio, si tuviéramos un polinomio con una potencia gigante (como \\(n^{100}\\)), el polinomio iría ganando por mucho porque empieza con una ventaja descomunal. Sin embargo, la exponencial funciona como un efecto multiplicativo en cadena. **Al final de la carrera, la exponencial siempre alcanzará y dejará comiendo polvo al polinomio de manera irreversible**.

Por esta razón, en el mundo de la informática, un programa con tiempo polinomial se puede usar perfectamente para procesar millones de datos, pero un programa con tiempo exponencial se vuelve imposible de ejecutar en la realidad porque tardaría más años que la edad del universo en terminar ante cualquier problema mediano.


- *Define qué es un 'algoritmo', asegurándote de mencionar sus características clave, y describe qué significa el 'Problema de Paro'formulado por Alan Turing.*- *Sintetiza la diferencia entre realizar un 'Análisis a Priori' y una 'Prueba a Posteriori' para medir la eficiencia de un algoritmo.*  
**RESPUESTA:**  

Un **algoritmo** es un conjunto finito de instrucciones precisas, concretas y no ambiguas diseñadas para resolver un problema mediante la transformación de datos de entrada en datos de salida. Para que un procedimiento sea considerado formalmente un algoritmo, debe cumplir con las siguientes características fundamentales:
*   **Entrada y Salida:** Debe tener definidos claramente los datos iniciales con los que trabajará y el resultado exacto que va a producir al finalizar.
*   **Precisión (No ambiguo):** Debe indicar explícitamente y sin ninguna duda el orden exacto en que se tienen que realizar los pasos.
*   **Finitud:** El procedimiento debe terminar necesariamente después de concluir un número finito de pasos.
*   **Determinismo:** Si se ejecuta varias veces con el mismo conjunto de datos de entrada, siempre debe pasar por los mismos estados y producir exactamente el mismo resultado final.
*   **Efectividad:** Cada paso o acción debe ser básico y realizable en un tiempo finito.
*   **Correctitud:** Debe resolver efectivamente el problema, garantizando encontrar la solución correcta para todas sus entradas.
*   **Generalidad:** Debe ser lo suficientemente general como para contemplar y responder correctamente ante cualquier posible caso de entrada dentro de las especificaciones del problema.

#### El Problema de Paro (*Halting Problem*) de Alan Turing
Formulado por el matemático Alan Turing en 1936, consiste en **determinar si cualquier algoritmo o programa de computadora llegará en algún momento a detenerse o si continuará ejecutándose indefinidamente** (como entrar en un ciclo infinito) al trabajar con una entrada dada.
Turing demostró matemáticamente la **insolubilidad de este problema** (clasificándolo como un problema *indecidible*). Esto significa que **es imposible que exista un algoritmo general capaz de examinar cualquier programa arbitrario junto con su entrada y dictaminar con certeza absoluta si terminará o no**. Si intentáramos resolverlo simplemente corriendo el programa y esperando a ver si termina, nunca tendríamos la certeza de si el programa está trabado en un bucle infinito o si se detendrá en el siguiente segundo de ejecución.

#### Diferencia entre "Análisis a Priori" y "Prueba a Posteriori"
La medición de la eficiencia (tiempo de ejecución y memoria) de un algoritmo se realiza bajo dos enfoques o fases de estudio bien diferenciadas:
*   **Análisis a Priori (Estudio Teórico):** Se realiza en la etapa de diseño de forma analítica, estudiando una representación de alto nivel del algoritmo (como pseudocódigo) sin necesidad de programarlo en una computadora. Consiste en obtener una **función matemática (como \\(T(n)\\))** que limita o acota teóricamente el tiempo de cálculo en función del tamaño de los datos de entrada. Su gran ventaja es que es **totalmente independiente** del ordenador, del lenguaje de programación, del compilador o de la habilidad del programador.
*   **Prueba a Posteriori (Estudio Empírico):** Se realiza en una fase experimental y requiere codificar, compilar y ejecutar el algoritmo en un ordenador real. Consiste en **recolectar estadísticas físicas reales de tiempo y espacio** consumidos por el programa mientras se ejecuta con conjuntos de datos de prueba limitados. Es **totalmente dependiente** de factores coyunturales y físicos: una computadora concreta, un compilador concreto, un lenguaje concreto y datos de prueba específicos.

- *Explica en qué consisten el peor caso, mejor caso y caso promedio al momento de evaluar el comportamiento de un algoritmo ante distintas entradas.*  
**RESPUESTA:**  
### Explicación Normal

En el análisis de algoritmos, las instrucciones que se ejecutan dependen de los datos concretos que se presenten como entrada. Por lo tanto, para un mismo tamaño de entrada \\(n\\), el tiempo de ejecución no es un valor único, sino que varía en un rango de valores limitado por un mínimo y un máximo: \\(T_{\min}(n) \le T(n) \le T_{\max}(n)\\).

Para evaluar este comportamiento de manera teórica, se definen tres escenarios:

#### 1. Peor Caso (*Worst Case*)
*   **En qué consiste:** Es la situación en la que el algoritmo realiza la **mayor cantidad de operaciones básicas** o consume el máximo tiempo posible para resolver un problema de tamaño \\(n\\).
*   **Definición matemática:** Se define formalmente como la función:
    \\[W(n) = \max \{t(I) \mid I \in D_n\}\\]
    donde \\(D_n\\) es el conjunto de todas las entradas posibles de tamaño \\(n\\), y \\(t(I)\\) es el número de operaciones con la entrada específica \\(I\\).
*   **Importancia:** Proporciona una **cota superior** y una garantía absoluta de que el algoritmo nunca tardará más que este límite bajo ninguna circunstancia. Es el análisis más común y prioritario en computación por su carácter conservador.

#### 2. Mejor Caso (*Best Case*)
*   **En qué consiste:** Es la situación en la que el algoritmo realiza el **mínimo número de operaciones elementales** o consume el menor tiempo posible para resolver un problema de tamaño \\(n\\). Representa la cota inferior \\(T_{\min}(n)\\).
*   **Importancia:** Generalmente se considera **engañoso** porque ocurre bajo condiciones sumamente optimistas o poco probables en la práctica. Sin embargo, es útil si este buen rendimiento se puede extender a entradas que estén "casi" en el mejor caso (como el ordenamiento por inserción, que es extremadamente rápido con arreglos casi ordenados).

#### 3. Caso Promedio (*Average Case*)
*   **En qué consiste:** Indica el **tiempo esperado o comportamiento típico** que tiene el algoritmo ante una entrada cualquiera de tamaño \\(n\\).
*   **Definición matemática:** Se define formalmente como:
    \\[A(n) = \sum_{I \in D_n} \Pr(I) t(I)\\]
    donde \\(\Pr(I)\\) es la probabilidad de que se presente la entrada específica \\(I\\).
*   **Importancia:** Es la medida de rendimiento **más realista y práctica** en el día a día. No obstante, es el análisis **más difícil de calcular** porque requiere asumir o conocer una distribución estadística de probabilidad para las entradas (por ejemplo, suponer que todas las entradas posibles son igualmente probables).

1. Implementa una versión recursiva para el cálculo del factorial ($n! = n \cdot (n-1)!$) y una versión iterativa para Fibonacci.

````python
def factorial_recursivo(n: int) -> int:
    # Validación para evitar valores negativos [4, 6]
    if n < 0:
        raise ValueError("No es posible calcular el factorial de números negativos [6]")

    # Caso base (solución trivial): 0! = 1 [2, 3, 5, 7]
    if n == 0:
        return 1

    # Caso recursivo: n * factorial(n - 1) [2, 3, 5, 8]
    else:
        return n * factorial_recursivo(n - 1)
````

````python
def fibonacci_iterativo(n: int) -> int:
    # Validación o caso base trivial para n <= 1 [13]
    if n <= 1:
        return n

    # Inicialización de los dos primeros términos [13]
    lofib = 0  # F(0)
    hifib = 1  # F(1)

    # Bucle iterativo desde 2 hasta n [13]
    for i in range(2, n + 1):
        x = lofib
        lofib = hifib
        hifib = x + lofib  # Se suma el término previo con el actual [13]

    return hifib
````

2. Utiliza la librería `time` o la función mágica `%timeit` para comparar el tiempo consumido (A posteriori) entre el Fibonacci recursivo y el iterativo para `n=35`.

````python
import time

def fibonacci_iterativo(n: int) -> int:
    if n <= 1:
        return n
    lofib, hifib = 0, 1
    for _ in range(2, n + 1):
        lofib, hifib = hifib, lofib + hifib
    return hifib

# Medición del método iterativo
start_iter = time.perf_counter()
res_iter = fibonacci_iterativo(35)
end_iter = time.perf_counter()
t_iter = end_iter - start_iter

print(f"Iterativo: {t_iter:.8f} seg (Resultado: {res_iter})")
````

````python
import time

def fibonacci_recursivo(n: int) -> int:
    if n <= 1:
        return n
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)

# Medición del método recursivo
start_rec = time.perf_counter()
res_rec = fibonacci_recursivo(35)
end_rec = time.perf_counter()
t_rec = end_rec - start_rec

print(f"Recursivo: {t_rec:.8f} seg (Resultado: {res_rec})")
````

Este cuadro compara de forma directa el comportamiento de las dos funciones en Python proporcionadas, explicando de manera analítica (*a priori*) y experimental (*a posteriori*) sus diferencias de rendimiento:

| Criterio | Versión Iterativa (`fibonacci_iterativo`) | Versión Recursiva (`fibonacci_recursivo`) |
| :--- | :--- | :--- |
| **Mecanismo de Control en el Código** | Utiliza un **bucle `for` único** que realiza exactamente \\(n-1\\) iteraciones simples. | Utiliza **dos llamadas recursivas simultáneas** en cada paso (`fibo(n-1) + fibo(n-2)`). |
| **Complejidad Temporal (A priori)** | **\\(O(n)\\) (Lineal)**. Pertenece a la familia de algoritmos polinomiales (en este caso, un polinomio de primer grado). | **\\(O(2^n)\\) o \\(O(\phi^n)\\) (Exponencial)**. Donde \\(\phi \approx 1.618\\) es la razón dorada. |
| **Redundancia de Cálculos** | **Nula.** En cada iteración calcula un nuevo término de Fibonacci apoyándose en los dos valores anteriores previamente guardados. | **Masiva.** Al no tener memoria, recalcula subproblemas idénticos millones de veces por separado (por ejemplo, para calcular \\(F(35)\\) recalcula \\(F(3)\\) millones de veces). |
| **Uso de Memoria en la Pila (*Stack*)** | **Constante (\\(O(1)\\)).** Solo utiliza dos variables en memoria (`lofib` e `hifib`) que se van actualizando en cada ciclo. | **Lineal (\\(O(n)\\)).** Almacena simultáneamente hasta \\(n\\) marcos de activación en la pila de llamadas del sistema antes de poder retornar un valor. |

| **Comportamiento Asintótico** | El tiempo crece en **proporción directa** al tamaño de \\(n\\). Si \\(n\\) se duplica, el número de sumas simplemente se duplica. | El tiempo **se multiplica** a medida que \\(n\\) crece, debido a que el árbol de llamadas recursivas se duplica en cada nivel. |
